## 1. Variables, Data Types & Strings
### **Concept Overview**
- **Dynamic Typing:** Python infers variable types at runtime—no explicit type declaration is needed.
- **Primitive Types:**
  - `int`: Whole numbers (e.g., `42`, `-7`)
  - `float`: Decimal numbers (e.g., `3.14`, `2.0`)
  - `bool`: Logical values (`True` or `False`)
  - `str`: Immutable sequences of Unicode characters
  - `NoneType`: Represents the absence of a value (`None`)
- **String Immutability:** Strings cannot be modified in place; string methods always return a *new* string.

```python
# 1. Variable Assignment & Dynamic Typing
student_name = "Aarav"          # str
age = 22                        # int
gpa = 3.85                      # float
is_enrolled = True              # bool
graduation_date = None          # NoneType

# Multiple assignment & unpacking
x, y, z = 10, 20, 30
print(f"Student: {student_name} | Age: {age} | GPA: {gpa:.1f} | Enrolled: {is_enrolled}")
print(f"Types -> age: {type(age).__name__}, gpa: {type(gpa).__name__}, grad: {type(graduation_date).__name__}")

# 2. Type Casting (Explicit Conversion)
raw_score = "94.5"
numeric_score = float(raw_score)
rounded_score = int(numeric_score)  # Truncates decimal part towards zero
print(f"\nCast '94.5' -> float: {numeric_score} -> int: {rounded_score}")

# 3. Essential String Operations & Slicing [start:stop:step]
course = "  Data Science & Machine Learning  "
clean_course = course.strip()                     # Removes leading/trailing whitespace
print(f"\nCleaned String : '{clean_course}'")
print(f"Upper Case     : '{clean_course.upper()}'")
print(f"Replace        : '{clean_course.replace('&', 'and')}'")
print(f"Split to List  : {clean_course.split(' ')}")
print(f"Slicing [0:12] : '{clean_course[0:12]}'")
print(f"Reverse [::-1] : '{clean_course[::-1]}'")
```

## 2. Core Data Structures
### **Comparison of Built-in Collections**

| Structure | Syntax | Ordered? | Mutable? | Allows Duplicates? | Common Use Case |
| :--- | :--- | :--- | :--- | :--- | :--- |
| **List** | `[a, b, c]` | Yes | Yes | Yes | Dynamic sequences of items |
| **Tuple** | `(a, b, c)` | Yes | **No** | Yes | Fixed records, coordinates, dict keys |
| **Dictionary** | `{'k': v}` | Yes (Python 3.7+) | Yes | Keys: **No**, Values: Yes | Fast $O(1)$ key-value lookups |
| **Set** | `{a, b, c}` | **No** | Yes | **No** | Deduplication & membership testing ($O(1)$) |

```python
# --- A. LISTS (Mutable, Ordered) ---
scores = [78, 85, 92, 64, 88]
scores.append(95)              # Add to end
scores.insert(1, 80)           # Insert at index 1
removed = scores.pop()         # Remove & return last item
scores.sort(reverse=True)      # Sort in-place descending
print("Sorted Scores:", scores)
print("Top 3 Scores (Slicing):", scores[:3])

# List Comprehension: [expression for item in iterable if condition]
high_scores_squared = [s**2 for s in scores if s >= 80]
print("High Scores Squared:", high_scores_squared)

# --- B. TUPLES (Immutable, Ordered) ---
dimensions = (1920, 1080)
width, height = dimensions     # Tuple unpacking
print(f"\nScreen Resolution: {width}x{height}")

# --- C. DICTIONARIES (Key-Value Mapping) ---
student = {
    "id": 101,
    "name": "Priya",
    "skills": ["Python", "SQL"],
    "Graduated": False
}
student["gpa"] = 3.9                           # Add/Update key
dept = student.get("department", "Unassigned") # Safe lookup with default value
print("\nStudent Dict Keys:", list(student.keys()))
print("Safe Lookup ('department'):", dept)

# Dictionary Comprehension
word_lengths = {w: len(w) for w in ["Data", "Science", "AI"]}
print("Word Lengths Dict:", word_lengths)

# --- D. SETS (Unique, Unordered Elements) ---
raw_tags = ["python", "ml", "data", "python", "ai", "ml"]
unique_tags = set(raw_tags)
required_tags = {"python", "sql", "ml"}
print("\nUnique Tags:", unique_tags)
print("Intersection (&):", unique_tags & required_tags)
print("Difference (-):", required_tags - unique_tags)
```

## 3. Control Flow & Loops
### **Concept Overview**
- **`if` / `elif` / `else`:** Executes conditional branches based on truthiness. Note that empty collections (`[]`, `{}`), `0`, `""`, and `None` evaluate to `False`.
- **`for` loops:** Iterate over any iterable (lists, strings, dictionaries, `range()`).
- **`while` loops:** Repeat as long as a condition remains `True`.
- **`break` vs `continue`:** `break` exits the loop immediately; `continue` skips the rest of the current iteration and jumps to the next.
- **`enumerate()` & `zip()`:** Built-in helpers for tracking indices and iterating over multiple sequences in parallel.

```python
# 1. Conditional Statements
marks = 82
if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 60:
    grade = "C"
else:
    grade = "F"
print(f"Marks: {marks} -> Grade: {grade}")

# Ternary Conditional Expression
status = "Pass" if marks >= 60 else "Fail"
print(f"Status: {status}")

# 2. For Loop with enumerate() and zip()
names = ["Alice", "Bob", "Charlie"]
test_scores = [91, 55, 84]

print("\n--- Student Evaluation Loop ---")
for idx, (name, score) in enumerate(zip(names, test_scores), start=1):
    if score < 60:
        print(f"#{idx}: {name} ({score}) -> Needs Retake (Skipping certificate)")
        continue
    print(f"#{idx}: {name} ({score}) -> Certificate Issued!")

# 3. While Loop with break
counter = 5
countdown = []
while counter > 0:
    countdown.append(counter)
    if counter == 2:
        break  # Early exit
    counter -= 1
print("\nWhile Loop exited early at 2:", countdown)
```

## 4. Functions & Functional Programming
### **Concept Overview**
- **Functions (`def`):** Reusable blocks of code that accept inputs (parameters) and return outputs.
- **Default Parameters:** Must follow non-default parameters (e.g., `def func(a, b=10):`).
- **`*args` & `**kwargs`:**
  - `*args` collects extra positional arguments as a **tuple**.
  - `**kwargs` collects extra keyword arguments as a **dictionary**.
- **`lambda` Functions:** Small, anonymous inline functions written as `lambda arguments: expression`.

```python
# 1. Standard Function with Docstring, Type Hints, and Default Argument
def calculate_compound_interest(principal: float, rate: float, years: int = 1) -> float:
    """Calculates final amount using annual compound interest."""
    return round(principal * ((1 + rate) ** years), 2)

print("1 Year (Default):", calculate_compound_interest(1000, 0.08))
print("5 Years (Keyword):", calculate_compound_interest(principal=1000, rate=0.08, years=5))

# 2. Flexible Arguments (*args and **kwargs)
def log_experiment(experiment_name: str, *metrics: float, **metadata: str) -> None:
    print(f"\nExperiment : {experiment_name}")
    print(f"Metrics    : {metrics} (Tuple of length {len(metrics)})")
    print(f"Metadata   : {metadata} (Dict)")

log_experiment("RandomForest_v1", 0.89, 0.92, 0.87, author="Rohan", dataset="Titanic")

# 3. Lambda Expressions with sorted(), map(), and filter()
products = [
    {"item": "Laptop", "price": 1200},
    {"item": "Mouse", "price": 25},
    {"item": "Monitor", "price": 300}
]

# Sort list of dicts by 'price' ascending using lambda
sorted_products = sorted(products, key=lambda x: x["price"])
print("\nSorted by Price:", sorted_products)

nums = [1, 2, 3, 4, 5, 6]
evens = list(filter(lambda n: n % 2 == 0, nums))
cubes = list(map(lambda n: n ** 3, evens))
print(f"Evens: {evens} -> Cubed: {cubes}")
```

## 5. Exception Handling
### **Concept Overview**
Real-world data pipelines encounter corrupt files, missing keys, and division by zero. Exception handling prevents your program from crashing.
- **`try`:** Code that might raise an error.
- **`except`:** Catches and handles specific exceptions (e.g., `ValueError`, `ZeroDivisionError`, `KeyError`). Always catch specific errors rather than a bare `except:`.
- **`else`:** Runs *only* if no exception was raised in the `try` block.
- **`finally`:** Runs *always*, regardless of whether an exception occurred (ideal for closing files or DB connections).
- **`raise`:** Manually triggers an exception when custom validation fails.

```python
def safe_divide_and_parse(numerator_str: str, denominator: float) -> float:
    try:
        num = float(numerator_str)
        if num < 0:
            raise ValueError("Numerator cannot be negative for this metric.")
        result = num / denominator
    except ValueError as ve:
        print(f"[ValueError Caught] Invalid value: {ve}")
        return 0.0
    except ZeroDivisionError:
        print("[ZeroDivisionError Caught] Cannot divide by zero!")
        return 0.0
    else:
        print(f"[Success] {num} / {denominator} = {result:.2f}")
        return result
    finally:
        print("  -> Execution of safe_divide_and_parse complete.")

# Test cases
safe_divide_and_parse("100", 4)       # Valid
safe_divide_and_parse("abc", 4)       # Triggers ValueError on float conversion
safe_divide_and_parse("50", 0)        # Triggers ZeroDivisionError
safe_divide_and_parse("-10", 2)       # Triggers custom raised ValueError
```

## 6. NumPy Essentials (Numerical Python)
### **Concept Overview**
- **`ndarray`:** NumPy's core N-dimensional array object. Unlike Python lists, NumPy arrays are stored in contiguous memory blocks and contain homogeneous data types, enabling C-speed computation.
- **Vectorization:** Applying operations to entire arrays at once without writing slow Python `for` loops.
- **Broadcasting:** NumPy's ability to perform arithmetic between arrays of different shapes (e.g., adding a 1D row vector to every row of a 2D matrix).
- **Boolean Masking:** Filtering arrays using conditional expressions like `arr[arr > 50]`.

```python
import numpy as np

# 1. Creating Arrays
arr_1d = np.array([10, 20, 30, 40, 50])
arr_2d = np.arange(1, 13).reshape(3, 4)   # 1 to 12 reshaped into 3 rows, 4 columns
zeros = np.zeros((2, 3))
lin = np.linspace(0, 1, 5)                # 5 evenly spaced points between 0 and 1

print("2D Matrix (3x4):\n", arr_2d)
print(f"Shape: {arr_2d.shape} | Dimensions: {arr_2d.ndim} | Dtype: {arr_2d.dtype}")
print("Linspace(0, 1, 5):", lin)

# 2. Indexing & Slicing 2D Arrays: arr[row_slice, col_slice]
print("\nElement at Row 1, Col 2 :", arr_2d[1, 2])
print("First 2 Rows, Last 2 Cols:\n", arr_2d[:2, 2:])

# 3. Vectorization & Broadcasting (No loops needed!)
prices = np.array([100.0, 250.0, 400.0, 80.0])
discounted = prices * 0.85 + 5.0          # Broadcasts scalar multiplication & addition
print("\nDiscounted + Shipping:", discounted)

# 4. Boolean Masking & np.where()
high_value_mask = prices >= 200
print("Prices >= 200:", prices[high_value_mask])
tier_labels = np.where(prices >= 200, "Premium", "Standard")
print("Tier Labels  :", tier_labels)

# 5. Axis Aggregations (axis=0 -> down columns, axis=1 -> across rows)
print("\nTotal Sum        :", arr_2d.sum())
print("Column Means (axis=0):", arr_2d.mean(axis=0))
print("Row Maxes (axis=1)   :", arr_2d.max(axis=1))
```

## 7. Pandas Essentials (Data Manipulation)
### **Concept Overview**
- **`Series`:** A 1D labeled array (like a single column in a table).
- **`DataFrame`:** A 2D labeled tabular structure with rows and heterogeneous columns.
- **`.loc` vs `.iloc`:**
  - **`.loc[row_label, col_label]`:** Label-based indexing (inclusive of the stop bound!).
  - **`.iloc[row_pos, col_pos]`:** Integer position-based indexing (exclusive of the stop bound, just like Python lists).
- **`groupby()`:** Implements the *Split-Apply-Combine* pattern for aggregating metrics across categories.

```python
import pandas as pd

# 1. Creating a Sample DataFrame
data = {
    "emp_id": [101, 102, 103, 104, 105, 106],
    "name": ["Asha", "Bikram", "Chitra", "Dev", "Esha", "Farhan"],
    "department": ["Engineering", "Sales", "Engineering", "HR", "Sales", "Engineering"],
    "salary": [95000, 62000, 110000, 58000, 71000, 88000],
    "experience_yrs": [4, 2, 7, 3, 4, 3]
}
df = pd.DataFrame(data)
print("--- Initial DataFrame ---")
print(df.head(3))

# 2. Selecting with .loc (Label-based) vs .iloc (Position-based)
print("\n.loc[0:2, ['name', 'salary']] (Inclusive of index 2!):\n", df.loc[0:2, ["name", "salary"]])
print("\n.iloc[0:2, 1:4] (Exclusive of index 2!):\n", df.iloc[0:2, 1:4])

# 3. Conditional Filtering & Creating Derived Columns
df["salary_per_yr_exp"] = (df["salary"] / df["experience_yrs"]).round(0)
senior_eng = df[(df["department"] == "Engineering") & (df["salary"] > 90000)]
print("\nFiltered (Engineering & Salary > 90k):\n", senior_eng[["name", "department", "salary"]])

# 4. GroupBy & Multi-Metric Aggregation
dept_summary = df.groupby("department").agg(
    headcount=("emp_id", "count"),
    avg_salary=("salary", "mean"),
    max_exp=("experience_yrs", "max")
).reset_index()
print("\nDepartment Summary (GroupBy):\n", dept_summary)
```

## 8. Basics of Data Cleaning
### **Concept Overview**
Raw datasets are rarely ready for analysis or machine learning. A standard data cleaning checklist includes:
1. **Removing Duplicates:** Using `df.duplicated()` and `df.drop_duplicates()`.
2. **String Standardization:** Stripping whitespace, standardizing case (`str.lower()`, `str.title()`), and removing currency symbols.
3. **Fixing Data Types:** Converting string numbers or dates into `float`, `int`, or `datetime` (`pd.to_numeric()`, `pd.to_datetime()`).
4. **Handling Missing Values (`NaN`):**
   - Drop rows/columns if missingness is minimal or structural (`df.dropna()`).
   - Impute numerical columns with **median** (robust to outliers) or **mean** (`df.fillna()`).
   - Impute categorical columns with the **mode** (most frequent value) or `'Unknown'`.
5. **Outlier Detection & Treatment (IQR Method):**
   - $IQR = Q_3 - Q_1$
   - Lower Bound $= Q_1 - 1.5 \times IQR$, Upper Bound $= Q_3 + 1.5 \times IQR$
   - Cap/clip extreme values using `.clip(lower, upper)` or filter them out.

```python
# Create a messy dataset with missing values, duplicates, dirty strings, and an outlier
messy_data = {
    "customer_id": [1, 2, 2, 3, 4, 5, 6, 7],
    "city": [" bengaluru ", "MUMBAI", "MUMBAI", "delhi", None, "Bengaluru", "Delhi ", "mumbai"],
    "age": [25, 32, 32, np.nan, 29, 41, 220, 35],          # 220 is a data entry outlier!
    "spend_usd": ["$120.50", "$85.00", "$85.00", "$210.00", "invalid", "$150.00", "$95.00", "$310.00"]
}
df_raw = pd.DataFrame(messy_data)
print("=== 1. RAW MESSY DATAFRAME ===")
print(df_raw)

# Step A: Remove exact duplicate rows
df_clean = df_raw.drop_duplicates().copy()

# Step B: Clean string columns (strip whitespace & title case)
df_clean["city"] = df_clean["city"].str.strip().str.title()

# Step C: Clean currency string & convert to float (errors='coerce' turns 'invalid' into NaN)
df_clean["spend_usd"] = (
    df_clean["spend_usd"]
    .str.replace("$", "", regex=False)
    .pipe(pd.to_numeric, errors="coerce")
)

# Step D: Handle Outliers in 'age' BEFORE imputing (cap using IQR or logical domain bounds)
Q1 = df_clean["age"].quantile(0.25)
Q3 = df_clean["age"].quantile(0.75)
IQR = Q3 - Q1
upper_bound = Q3 + 1.5 * IQR
df_clean["age"] = df_clean["age"].clip(upper=upper_bound)

# Step E: Impute Missing Values (Median for numeric, Mode for categorical)
df_clean["age"] = df_clean["age"].fillna(df_clean["age"].median())
df_clean["spend_usd"] = df_clean["spend_usd"].fillna(df_clean["spend_usd"].median())
df_clean["city"] = df_clean["city"].fillna(df_clean["city"].mode()[0])

print("\n=== 2. CLEANED DATAFRAME ===")
print(df_clean.reset_index(drop=True))
print("\nMissing Values Count After Cleaning:\n", df_clean.isna().sum())
```

## 9. Machine Learning Preprocessing & Train-Test Split
### **Concept Overview**
- **Why Split into Train and Test?**
  - We train our model on the **Training Set** (typically 70%–80% of the data) and evaluate its generalization on unseen data using the **Test Set** (20%–30%).
- **Key Parameters of `train_test_split`:**
  - `test_size`: Proportion of data reserved for testing (e.g., `0.2` for 20%).
  - `random_state`: Fixes the random seed so the exact split is reproducible every time.
  - `stratify=y`: Preserves the exact class ratio of the target variable `y` in both train and test sets (crucial for imbalanced classification!).
- **Golden Rule — Preventing Data Leakage:**
  - Always split your data **BEFORE** fitting scalers (`StandardScaler`) or imputers!
  - Use `.fit_transform(X_train)` on the training set, and **only `.transform(X_test)`** on the test set so test-set statistics never leak into training.

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer

# 1. Create a synthetic dataset for binary classification (Customer Churn)
np.random.seed(42)
n_samples = 100
ml_df = pd.DataFrame({
    "tenure_months": np.random.randint(1, 60, size=n_samples),
    "monthly_bill": np.random.uniform(20.0, 150.0, size=n_samples).round(2),
    "contract_type": np.random.choice(["Monthly", "One-Year", "Two-Year"], size=n_samples),
    "churned": np.random.choice([0, 1], size=n_samples, p=[0.8, 0.2])  # Imbalanced: 80% No, 20% Yes
})

# 2. Separate Features (X) and Target (y)
X = ml_df.drop(columns=["churned"])
y = ml_df["churned"]

# 3. Perform Stratified Train-Test Split (80% Train, 20% Test)
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y  # Ensures both train & test have ~20% churn rate
)

print(f"Total Rows: {len(ml_df)} | X_train: {X_train.shape} | X_test: {X_test.shape}")
print(f"Train Churn Rate: {y_train.mean():.2f} | Test Churn Rate: {y_test.mean():.2f}")

# 4. Feature Scaling & Categorical Encoding WITHOUT Data Leakage
numeric_cols = ["tenure_months", "monthly_bill"]
categorical_cols = ["contract_type"]

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_cols),
        ("cat", OneHotEncoder(drop="first", sparse_output=False), categorical_cols)
    ]
)

# FIT ONLY ON TRAIN, then transform both Train and Test!
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

# Convert back to DataFrame for easy inspection
feature_names = preprocessor.get_feature_names_out()
df_train_scaled = pd.DataFrame(X_train_processed, columns=feature_names)
print("\n--- Processed X_train (First 3 Rows) ---")
print(df_train_scaled.head(3).round(3))
```
